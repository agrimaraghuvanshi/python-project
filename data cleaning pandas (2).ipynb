```python
import pandas as pd
#Read csv:
df = pd.read_csv("data.csv")
#Convert date to datetime
df["Date"] = pd.to_datetime(df["Date"])
#Make some data wrong
df.loc[2,"Pulse"] = -10
df.loc[5,"Calories"] = 500
df.loc[7,"Maxpulse"] = -100
df.loc[4,"Duration"] = 150
#Remove rows where duration>120
for x in df.index:
    if df.loc[x,"Duration"]>120:
        df.drop(x,inplace=True)

#Mean
print("Mean:")
print(df[["Duration","Pulse","Maxpulse","Calories"]].mean())

#Median
print("\nMedian:")
print(df[["Duration","Pulse","Maxpulse","Calories"]].median())

#Mode
print("\nMode:")
print(df[["Duration","Pulse","Maxpulse","Calories"]].mode())
```

```python
import pandas as pd
#Read csv:
df = pd.read_csv("data.csv")
#Convert date to datetime
df["Date"] = pd.to_datetime(df["Date"])
#Create a duplicate row
df.loc[10] = df.loc[2]
print("Before removing duplicates:")
print(df)
#find duplicates
print("\nDuplicate rows:")
print(df[df.duplicated()])
#Remove duplicates
df.drop_duplicates(inplace=True)
print("\nAfter removing duplicates:")
print(df)
```
